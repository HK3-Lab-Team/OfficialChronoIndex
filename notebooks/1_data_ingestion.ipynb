## Zhao (2023)

```python
zhao = datasets["zhao"]
```

```python
print("metadata:", zhao.metadata.shape, zhao.metadata.columns)
print("cgm_data:", zhao.cgm_data.shape, zhao.cgm_data.columns)
print("joined_data:", zhao.joined_data.shape, zhao.joined_data.columns)
```

```python
print(zhao.metadata.head())
print(zhao.cgm_data.head())
print(zhao.joined_data.head())
```

```python
# Rebuild Methods (for understanding internals)
cgm_rebuilt = zhao.construct_cgm_dataframe()
joined_rebuilt = zhao.construct_joined_dataframe()
print("rebuilt cgm_data shape:", cgm_rebuilt.shape)
print("rebuilt joined_data shape:", joined_rebuilt.shape)
```

Subject-Level Access

Use `ids` + `get_subject_data` to inspect one subject in detail.

```python
sid = zhao.ids[0]
subj = zhao.get_subject_data(sid)
print(subj.metadata)
```

```python
print("Subject CGM rows:", subj.cgm_readings.height)
subj.cgm_readings.head()
```

```python
subj.cgm_readings.select(
    pl.len().alias("n_rows"),
    pl.col("Date").min().alias("start"),
    pl.col("Date").max().alias("end"),
    pl.col("CGM").mean().alias("mean_cgm"),
    pl.col("CGM").min().alias("min_cgm"),
    pl.col("CGM").max().alias("max_cgm"),
)
```

CGM Representation Round-Trip

`cgm_to_single_row()` converts a subject to list-columns; `single_row_to_cgm()` restores long format.

```python
single = subj.cgm_to_single_row()
roundtrip = subj.single_row_to_cgm(single)
print("single-row shape:", single.shape)
print("roundtrip shape:", roundtrip.shape)
roundtrip.head()
```
